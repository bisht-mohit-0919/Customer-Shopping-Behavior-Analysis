```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

df.isnull().sum()

## Change All the columns to lowercase and replace space with hyphen

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns= {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

## CREATE A NEW COLUMNS FOR AGE GROUP

```python
labels = ['Young Adult', 'Adult', 'Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels)
```

```python
df[['age','age_group']]
```

### Create a new columns for purchase_frequency_days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually' : 365,
    'Every 3 months': 90
    
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']]
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

## DROP THE PROMO CODE USED COLUMNS BECAUSE ITS CONTAIN THE SAME INFORMATION THAT DISCOUNT APPLIED CONTAIN

```python
df = df.drop('promo_code_used', axis =1)
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# 1. Apna connection URL safe tareeke se banayein (Special character password ke liye best hai)
url_object = URL.create(
    "mysql+mysqlconnector",
    username="root",
    password="Mohit@7065",
    host="localhost",
    database="customer_behaviour_analysis"
)

# 2. Engine banayein (Yeh Pandas ko SQL se jodne ka pull/bridge hai)
engine = create_engine(url_object)

try:
    # 3. DataFrame (df) ko SQL database mein bhejein
    # 'customer_data' aapki nayi table ka naam hoga SQL mein (Aap isse badal sakte hain)
    # if_exists='replace' ka matlab hai ki agar table pehle se hai toh purani hata kar nayi daal de (Aap 'append' bhi use kar sakte hain)
    # index=False ka matlab hai ki df ke index (0,1,2..) ko SQL mein alag column nahi banayega
    
    df.to_sql(name='customer_data', con=engine, if_exists='replace', index=False)
    
    print("Badhai ho! Aapka DataFrame successfully MySQL mein transfer ho gaya hai! 🚀")

except Exception as e:
    print(f"Koi error aayi: {e}")
```
